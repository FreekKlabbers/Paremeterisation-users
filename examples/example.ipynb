Parameterisation - Example usage
=======================================

The parameterisation is function defined to go from the voltage to the current. This is represented as follows:

$$I = f(\eta, \frac{L_P}{\lambda_D},\delta, C_P)$$
where $I$ is the current, $f()$ is the function, $\eta$ is the normalised probe voltage, $\frac{L_P}{\lambda_D}$ is ????, $\delta$ is the collisionality and $C_P$ is the probe geometry. For a full recap, reference the paper.

The following steps have been designed to use the usage of this paremeterisation as easy as possible.

# Installation
To use this the Python package will need to be installed. This can be done locally or through PyPi??? (If we want this)

For a local installation follow the following Steps:

```bash
git clone [Repo name]
cd [Repo name]
pip install ParaPro --extra-index-url https://download.pytorch.org/whl/cpu
```
Which will install ParaPro in editable mode with CPU torch (which is a lot smaller). The GPU torch version will also work, its just overkill for an inference only application.

## Usage of installation
Once installed, usage is as easy as importing either of the functions which you want to use either `NN_predict` or `SR_predict`.

````python
from ParaPro import NN_predictor, SR_predict
````

## Example settings
Below we give a set of example settings which you may use.

````python
eta = 1.060949
R = 10  # we use R to replace the fraction of L_P and lambda_D
delta = 0.001
Cp = 6.32

X_in = eta, R, delta, Cp
````

## Usage of the Symbolic Regression Formula
This function is implemented fully in numpy.

````python
SR_predict(X_in)
````

# Usage of the Neural Network model
This class with predict function gets the machine learning model, sets it up and then uses it to predict.

If its the first run of the code, it will download the latest weights from Hugging Face.


````python
model = NN_predictor()
model.predict([X_in])
````
